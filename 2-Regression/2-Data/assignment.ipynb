```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pumpkins = pd.read_csv('../data/US-pumpkins.csv')

pumpkins = pumpkins[pumpkins['Package'].str.contains('bushel',case=True,regex=True)]
pumpkins.head()
```

```python
pumpkins.isna().sum()
```

```python
pumpkins.info()
```

```python
plt.hist(pumpkins['Low Price'],bins=20)
plt.title('Distribution of Low Price')
plt.ylabel('Low Price')

plt.show()
```

```python
sns.histplot(pumpkins['Low Price'],bins=20)
plt.title('Distribution of Low Price')
plt.ylabel('Low Price')
```

```python
plt.scatter(pumpkins['Low Price'],pumpkins['High Price'])
plt.show()
```

```python
sns.scatterplot(data=pumpkins,x='Low Price',y='High Price')
```

```python
sns.boxplot(data=pumpkins,x='Type',y='Low Price')
```
