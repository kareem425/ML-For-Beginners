```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd 
from sklearn import datasets , linear_model,model_selection
```

```python
x,y = datasets.load_linnerud(return_X_y=True)

print(x.shape)
print(x[0])
```

```python
x = x[:,1]
print(x.shape)

x=x.reshape((-1,1))
print(x.shape)
```

```python
y = y[:,1]
print(y.shape)

```

```python
x_tr,x_test,y_tr,y_test = model_selection.train_test_split(x,y,test_size=0.33)
```

```python
model = linear_model.LinearRegression()
model1 = model.fit(x_tr,y_tr)
```

```python
y_pred = model1.predict(x_test)
```

```python
print(x_test.shape)
print(y_test.shape)
```

```python
#creat the scatter
plt.scatter(x_test,y_test,color='black')

#plot the predictions
plt.plot(x_test,y_pred,color = 'blue',linewidth = 3)

plt.xlabel('situps')
plt.ylabel('waist')
plt.title('A Graph Plot Showing Situps & Waist ')

plt.show()
```
